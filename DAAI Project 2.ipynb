```python
#Flaging the high adj close price using the IDR method
Q1 = df2['adj_close'].quantile(0.25)
Q3 = df2['adj_close'].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

df2['adj_close_flag'] = df2['adj_close'] > upper_limit
```

***Part 2 — Create Useful Features***


```python
#Sort by ticker and date first
df2 = df2.sort_values(['ticker', 'date'])
```

```python
#Create price_change colume which is change in closing price from the previous trading day
df2['price_change'] = df2.groupby('ticker')['close'].diff()
```

```python
#show teh output
print(df2[['ticker', 'date', 'close', 'price_change']].head(10))
```

The NAN for the first  dates is because there is no previous trading day for the first observation of each ticker
Positive number implies a price increase
Negative number implies a price decrease

```python
# Create a daily return
df2['daily_return'] = df2.groupby('ticker')['close'].pct_change() * 100
```

The stockes are group per ticker to ensures that each is calculated separately for each stock.

```python
print(df2[['ticker', 'date', 'close', 'price_change','daily_return']].head(20))
```

```python
#close_lag_1
df2['close_lag_1'] = df2.groupby('ticker')['close'].shift(1)
```

```python
print(df2[['ticker', 'date', 'close', 'price_change','daily_return', 'close_lag_1']].head(20))
```

```python
#Create the 5-day moving average
df2['ma_5'] = (
    df2.groupby('ticker')['close']
       .rolling(window=5)
       .mean()
       .reset_index(level=0, drop=True)
)
```

```python
#Create the 20-day moving average
df2['ma_20'] = (
    df2.groupby('ticker')['close']
       .rolling(window=20)
       .mean()
       .reset_index(level=0, drop=True)
)
```

```python
print(
    df2[['ticker', 'date', 'close', 'ma_5', 'ma_20']].head(30)
)
```

```python
#Create volatility_20 using the dairly return already created,. Note that Volatility measures how much a stock's returns have been fluctuating
df2['volatility_20'] = (
    df2.groupby('ticker')['daily_return']
       .rolling(window=20)
       .std()
       .reset_index(level=0, drop=True)
)
```

```python
print(
    df2[['ticker', 'date', 'close', 'volatility_20']].head(30)
)
```

```python
#Additional feature: high_low_range
df2['high_low_range'] = df2['high'] - df2['low']
```

I selected high_low_range because it measures the amount of price movement within each trading day. A larger range indicates greater intra day price movement, which can provide additional information about stock behaviour and market activity.

A larger range means the stock had more price movement during that trading day.

```python
#pring the result
print(
    df2[
        ['ticker', 'date', 'close', 'daily_return',
         'volatility_20', 'high_low_range']
    ].head(30)
)
```

```python
#Final dataset features
final_features = [
    'ticker',
    'open',
    'close',
    'adj_close',
    'low',
    'high',
    'volume',
    'date',
    'price_change',
    'daily_return',
    'close_lag_1',
    'ma_5',
    'ma_20',
    'volatility_20',
    'high_low_range',
    'industry'
]

# Add the one-hot encoded exchange and sector columns
encoded_columns = [
    col for col in df2.columns
    if col.startswith('exchange_') or col.startswith('sector_')
]

final_features = final_features + encoded_columns

final_df = df2[final_features].copy()

print(final_df.head())
print(final_df.shape)
```
